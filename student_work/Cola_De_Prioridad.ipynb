```python
from goodrich.ch09.unsorted_priority_queue import *

class StackPQ():

    def __init__(self):
        self.data = UnsortedPriorityQueue()
        self.key = 0

    def __len__(self):
        return len (self.data)
    def is_empty(self):
        return len (self.data)==0
    def push(self,e):
        self.data.add(self.key,e)
        self.key -= 1
    def top (self):
        k,v = self.data.min()
        return v
    def pop (self):
        if len(self.data)==0:
            raise Empty("El stack esta vacio")
        k,v = self.data.remove_min()
        return v

```
